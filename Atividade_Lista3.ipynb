##4. Lista de números: pares e ímpares


Leia 10 números inteiros e armazene todos em uma lista. Depois das 10 leituras, percorra a lista e
crie duas novas listas: uma apenas com os pares e outra apenas com os ímpares. Mostre as três
listas.

```python
numeros = []

for i in range(10):
    numero = int(input(f"Digite o {i + 1}º número inteiro: "))
    numeros.append(numero)

pares = []
impares = []

for numero in numeros:
    if numero % 2 == 0:
        pares.append(numero)
    else:
        impares.append(numero)

print("Números:",numeros)
print("Pares:",pares)
print("Ímpares:",impares)
```

##5. Maior e menor da lista
Leia 8 números inteiros, armazene-os em uma lista e determine o maior e o menor valor sem usar
max() nem min(). Informe também as posições em que eles aparecem pela primeira vez.

```python
numeros = []

for i in range(8):
    valor = int(input(f"Digite o {i + 1}º número inteiro: "))
    numeros.append(valor)

maior_valor = numeros[0]
pos_maior = 0

menor_valor = numeros[0]
pos_menor = 0

for i in range(1, len(numeros)):
    if numeros[i] > maior_valor:
        maior_valor = numeros[i]
        pos_maior = i

    if numeros[i] < menor_valor:
        menor_valor = numeros[i]
        pos_menor = i

print(f"\nLista de números: {numeros}")
print(f"Maior valor: {maior_valor} (primeira aparição no índice: {pos_maior})")
print(f"Menor valor: {menor_valor} (primeira aparição no índice: {pos_menor})")
```

##6. Nomes sem repetição

Leia nomes até que o usuário digite FIM. Armazene somente nomes que ainda não estejam na
lista. Ao final, mostre a lista e a quantidade de nomes diferentes cadastrados.

```python
nomes = []

while True:
    nome = input("Digite um nome (ou FIM para encerrar): ")

    if nome.upper() == "FIM":
        break

    ja_existe = False
    for n in nomes:
        if n == nome:
            ja_existe = True
            break

    if ja_existe == False:
        nomes.append(nome)

print("\nLista de nomes cadastrados:")
print(nomes)
print(f"Quantidade de nomes diferentes: {len(nomes)}")
```

##7. Dias da semana em uma tupla
Crie uma tupla contendo os sete dias da semana, em ordem. Leia um número de 1 a 7 e mostre o
dia correspondente. Caso o número esteja fora desse intervalo, informe que a opção é inválida.

```python
dias_semana = ("Domingo", "Segunda-feira", "Terça-feira", "Quarta-feira", "Quinta-feira", "Sexta-feira", "Sábado")

numero = int(input("Digite um número de 1 a 7: "))

if numero >= 1 and numero <= 7:
    dia = dias_semana[numero - 1]
    print(f"O dia correspondente é: {dia}")
else:
    print("Opção inválida! Digite um número entre 1 e 7.")
```

##8. Estatísticas de uma tupla
Leia 6 números inteiros, forme uma tupla com eles e, depois, informe: quantas vezes o número 5
apareceu, a posição da primeira ocorrência de 5 (se existir) e quantos valores são pares.

```python
numeros = []
for i in range(6):
    num = int(input(f"Digite o {i+1}º número inteiro: "))
    numeros.append(num)

tupla_numeros = tuple(numeros)

qtd_cinco = tupla_numeros.count(5)

if 5 in tupla_numeros:
    posicao_cinco = tupla_numeros.index(5)
    info_posicao = f"O número 5 aparece pela primeira vez no índice {posicao_cinco} (posição {posicao_cinco + 1})."
else:
    info_posicao = "O número 5 não foi digitado."

qtd_pares = 0
for elemento in tupla_numeros:
    if elemento % 2 == 0:
        qtd_pares += 1

print("\n--- Estatísticas ---")
print(f"Tupla formada: {tupla_numeros}")
print(f"O número 5 apareceu {qtd_cinco} vez(es).")
print(info_posicao)
print(f"Quantidade de números pares: {qtd_pares}")
```

##9. Dicionário de aluno

Cadastre nome, turma, nota1 e nota2 de um aluno em um dicionário. Calcule a média, acrescente
ao próprio dicionário as chaves media e situacao e mostre todos os dados.

 Use: aprovado >= 7;
recuperação >= 5; reprovado < 5.

```python
aluno = {}

aluno['nome'] = input("Digite o nome do aluno: ")
aluno['turma'] = input("Digite a turma: ")
aluno['nota1'] = float(input("Digite a primeira nota: "))
aluno['nota2'] = float(input("Digite a segunda nota: "))

aluno['media'] = (aluno['nota1'] + aluno['nota2']) / 2

if aluno['media'] >= 7:
    aluno['situacao'] = "Aprovado"
elif aluno['media'] >= 5:
    aluno['situacao'] = "Recuperação"
else:
    aluno['situacao'] = "Reprovado"

print("\n--- Dados do Aluno ---")
print(f"Nome: {aluno['nome']}")
print(f"Turma: {aluno['turma']}")
print(f"Nota 1: {aluno['nota1']}")
print(f"Nota 2: {aluno['nota2']}")
print(f"Média: {aluno['media']:.2f}")
print(f"Situação: {aluno['situacao']}")
```

##11. Frequência das palavras
Leia 8 palavras e use um dicionário para contar quantas vezes cada palavra foi digitada. Ao final,
mostre cada palavra e sua quantidade

```python
frequencia = {}

for i in range(8):
    palavra = input(f"Digite a {i+1}ª palavra: ").strip().lower()

    if palavra in frequencia:
        frequencia[palavra] += 1
    else:
        frequencia[palavra] = 1

print("\n--- Frequência das Palavras ---")
for palavra, quantidade in frequencia.items():
    print(f"{palavra}: {quantidade}")
```

##12. Estoque e atualização
Crie um dicionário com 4 produtos e suas quantidades em estoque. Leia o nome de um produto e
uma quantidade vendida. Se o produto existir e houver estoque suficiente, atualize o dicionário;
caso contrário, informe o problema. Mostre o estoque final.

```python
estoque = {
    "arroz": 50,
    "feijao": 30,
    "leite": 20,
    "cafe": 15
}

produto = input("Digite o nome do produto vendido: ").lower()
quantidade_vendida = int(input("Digite a quantidade vendida: "))

if produto in estoque:
    if estoque[produto] >= quantidade_vendida:
        estoque[produto] -= quantidade_vendida
        print(f"Venda realizada com sucesso! Estoque atualizado de {produto}: {estoque[produto]}")
    else:
        print(f"Estoque insuficiente. Quantidade disponível de {produto}: {estoque[produto]}")
else:
    print(f"O produto '{produto}' não está cadastrado no estoque.")

print("\n--- Estoque Final ---")
for prod, qtd in estoque.items():
    print(f"{prod}: {qtd}")
```

##13. Cadastro de vários alunos
Cadastre 5 alunos. Para cada aluno, leia nome e duas notas, calcule a média e armazene nome,
notas e média em um dicionário. Cada dicionário deve ser adicionado a uma lista. Ao final, mostre
nome e média de todos os alunos.

```python
turma = []

for i in range(5):
    print(f"\n--- Cadastro do {i+1}º Aluno ---")
    aluno = {}
    aluno["nome"] = input("Digite o nome do aluno: ")
    aluno["nota1"] = float(input("Digite a primeira nota: "))
    aluno["nota2"] = float(input("Digite a segunda nota: "))
    aluno["media"] = (aluno["nota1"] + aluno["nota2"]) / 2

    turma.append(aluno)

print("\n--- Resultado Final ---")
for aluno in turma:
    print(f"Aluno: {aluno['nome']} | Média: {aluno['media']:.2f}")
```

##14. Catálogo com código imutável
Cadastre 5 produtos usando tuplas no formato (codigo, nome, preco) e armazene essas tuplas em
uma lista. Depois, leia um código e procure o produto correspondente, mostrando nome e preço.
Informe quando o código não for encontrado.

```python
produtos = []

for i in range(5):
    print(f"\n--- Cadastro do {i+1}º Produto ---")
    codigo = int(input("Digite o código do produto: "))
    nome = input("Digite o nome do produto: ")
    preco = float(input("Digite o preço do produto: "))

    produto = (codigo, nome, preco)
    produtos.append(produto)

codigo_busca = int(input("\nDigite o código do produto para buscar: "))

encontrado = False

for produto in produtos:
    if produto[0] == codigo_busca:
        print(f"\nProduto encontrado: {produto[1]}")
        print(f"Preço: R$ {produto[2]:.2f}")
        encontrado = True
        break

if not encontrado:
    print("\nCódigo não encontrado no catálogo.")
```

##15. Mini sistema de estoque

Construa um pequeno cadastro de 5 itens de estoque. Cada item deve ser um dicionário com
codigo, nome, quantidade e preco. Armazene os dicionários em uma lista. Depois:

 (a) mostre
todos os itens;

(b) calcule o valor total do estoque;

 (c) identifique o item de maior valor em estoque
(quantidade × preço);

 (d) crie uma tupla resumo contendo (quantidade_de_itens_cadastrados,
valor_total_estoque).

```python
estoque = []

for i in range(5):
    print(f"\n--- Cadastro do {i+1}º Item ---")
    codigo = int(input("Digite o código: "))
    nome = input("Digite o nome: ")
    quantidade = int(input("Digite a quantidade: "))
    preco = float(input("Digite o preço unitário: "))

    item = {
        "codigo": codigo,
        "nome": nome,
        "quantidade": quantidade,
        "preco": preco
    }

    estoque.append(item)

print("\n--- Todos os Itens Cadastrados ---")
for item in estoque:
    print(f"Código: {item['codigo']} | Nome: {item['nome']} | Qtd: {item['quantidade']} | Preço: R$ {item['preco']:.2f}")

valor_total_estoque = 0.0
maior_valor_item = -1.0
item_maior_valor = None

for item in estoque:
    valor_item = item["quantidade"] * item["preco"]
    valor_total_estoque += valor_item

    if valor_item > maior_valor_item:
        maior_valor_item = valor_item
        item_maior_valor = item

print(f"\nValor total do estoque: R$ {valor_total_estoque:.2f}")
print(f"Item com maior valor total em estoque: {item_maior_valor['nome']} (R$ {maior_valor_item:.2f})")

resumo = (len(estoque), valor_total_estoque)
print(f"\nTupla Resumo (Qtd de Itens, Valor Total): {resumo}")
```

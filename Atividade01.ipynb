```python
operacao=str(input("Digite uma operação: '+', '-': "))
num1=int(input("Digite o primeiro número: "))
num2=int(input("Digite o segundo número: "))
print("O resultado é: ")
if operacao=='+':
  print(num1 + num2)
elif operacao=='-':
  print(num1 - num2)
```

##4. Estacionamento

Um estacionamento cobra RS 8,00 por até 2 horas. Acima de 2 e até 5 horas, cobra RS 8,00 mais
R$ 3,00 por hora adicional.

Acima de 5 horas, cobra RS 17,00 mais R$ 2,00 por hora que
ultrapassar 5.

Leia a quantidade inteira de horas e calcule o valor a pagar.

```python
horas = float(input("Digite a quantidade de horas: "))
if horas <= 2:
  valor_final = 8
elif horas > 2 and horas <=5:
  valor_final = 8 + (horas - 2) * 3
elif horas > 5:
  valor_final = 17 + (horas - 5) * 2


print ("O valor que você vai pagar é:\n R$",valor_final)
```

##5. Máquina de venda de ingresso
Um ingresso custa RS 25,00.

Solicite valores inteiros ao usuário até que o total inserido seja
suficiente para a compra.

Ao final, informe o total inserido e o troco.

```python
ingresso = 25
total_inserido = 0

print(f"O ingresso custa R${ingresso}")

while total_inserido < ingresso:
  falta = ingresso - total_inserido
  print(f"Faltam R$ {falta} para o ingresso")

  valor = int(input("Digite um valor inteiro: "))
  total_inserido += valor

troco =  total_inserido - valor

print(f"Total inserido: R$ {total_inserido}")
print(f"Troco: R$ {troco}")
```

##6. Maior valor e sua posição
Leia 8 números inteiros.

Informe o maior valor e a posição em que ele apareceu pela primeira
vez

```python
maior = int(input("Digite o 1⁰ número: "))
posicao_maior = 1

for i in range (2,9):
  numero = int(input(f"Digite o {i}⁰ número: "))

  if numero > maior:
        maior = numero
        posicao_maior = i

print(f"O maior valor é {maior} e apareceu na posição {posicao_maior}")
```

##7. Segundo maior valor
Leia 8 números inteiros distintos e informe o maior e o segundo maior. Não utilize max(), sort()
ou listas.

```python
num1 = int(input("Digite o 1⁰ número: "))
num2 = int(input("Digite o 2⁰ número: "))

if num1 > num2:
  maior = num1
  segundo_maior = num2
else:
  maior = num2
  segundo_maior = num1

for i in range(3, 9):
    numero = int(input(f"Digite o {i}º número: "))

    if numero > maior:
       segundo_maior = maior
       maior = numero
    elif numero > segundo_maior:
       segundo_maior = numero

print(f"Maior valor: {maior}")
print(f"Segundo maior valor: {segundo_maior}")
```

##8. Detectando repetição consecutiva
Leia 8 números inteiros e informe se em algum momento foram digitados dois valores iguais
consecutivamente. Não é necessário informar quais valores foram iguais.

```python
anterior = int(input("Digite o 1º número: "))
posicao_repeticao = 0

for i in range(2, 9):
    atual = int(input(f"Digite o {i}º número: "))

    if atual == anterior:
        posicao_repeticao = i
        break

    anterior = atual

if posicao_repeticao > 0:
    print("Foram digitados valores iguais consecutivamente!")
else:
    print("Não houve valores iguais digitados consecutivamente.")
```

##9. Controle de saldo
O programa começa com saldo de R$ 500,00. Serão realizadas 6 operações. Em cada uma, leia
o tipo (1 = depósito; 2 = saque) e o valor.
Depósitos aumentam o saldo. Saques só podem
ocorrer se houver saldo suficiente; caso contrário, mostre 'Saldo insuficiente'. Ao final, mostre o
saldo.

```python
saldo = 500

print(f"Saldo inicial: R$ {saldo}")


for i in range(1, 7):
    print(f"\nOperação {i} de 6:")
    tipo = int(input("Informe o tipo (1 = Depósito | 2 = Saque): "))
    valor = int(input("Informe o valor: R$ "))

    if tipo == 1:
        saldo += valor
        print(f"Depósito de R$ {valor} realizado com sucesso.")
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
            print(f"Saque de R$ {valor} realizado com sucesso.")
        else:
            print("Saldo insuficiente")
    else:
        print("Tipo de operação inválido!")

print("\n" + "=" * 30)
print(f"Saldo final: R$ {saldo}")
```

##10. Caixa eletrônico simplificado
Leia um valor inteiro positivo e informe a menor quantidade de cédulas para representá-lo,
usando R$ 100, 50, 20, 10, 5, 2 e 1.

```python
valor = int(input("Digite o valor a ser sacado: R$ "))

cedulas = [100, 50, 20, 10, 5, 2, 1]

print(f"\nValor informado: R$ {valor}")
print("Cédulas entregues:")

for cedula in cedulas:
    quantidade = valor // cedula
    valor = valor % cedula

    if quantidade > 0:
        print(f"{quantidade} nota(s) de R$ {cedula}")
```

##11. Número primo
Leia um inteiro maior que 1 e informe se ele é primo.

```python
numero = int(input("Digite um número inteiro maior que 1: "))

if numero <= 1:
    print("O número deve ser maior que 1.")
else:
    divisores = 0

    for i in range(2, numero):
        if numero % i == 0:
            divisores += 1
            break

    if divisores == 0:
        print(f"O número {numero} é primo.")
    else:
        print(f"O número {numero} não é primo.")
```

##12. Divisores de um número
Leia um inteiro positivo N, mostre todos os seus divisores positivos e informe quantos divisores
ele possui.

```python
n = int(input("Digite um número inteiro positivo: "))

if n <= 0:
    print("O número deve ser positivo.")
else:
    total_divisores = 0

    print(f"Divisores positivos de {n}:")
    for i in range(1, n + 1):
        if n % i == 0:
            print(i)
            total_divisores += 1

    print(f"O número {n} possui {total_divisores} divisor(es).")
```

##13. Números primos em um intervalo
Leia A e B positivos, com A < B, mostre os primos entre A e B (inclusive) e informe quantos foram
encontrados.

```python
a = int(input("Digite o valor de A (positivo): "))
b = int(input("Digite o valor de B (positivo, maior que A): "))

if a <= 0 or b <= 0 or a >= b:
    print("Os valores devem ser positivos e A deve ser menor que B.")
else:
    quantidade_primos = 0

    print(f"Números primos entre {a} e {b}:")
    for numero in range(a, b + 1):
        if numero > 1:
            divisores = 0
            for i in range(2, numero):
                if numero % i == 0:
                    divisores += 1
                    break

            if divisores == 0:
                print(numero)
                quantidade_primos += 1

    print(f"Total de números primos encontrados: {quantidade_primos}")
```

##14. Jogo com limite de tentativas
Considere o número secreto 37. O usuário terá no máximo 5 tentativas. Após cada erro, informe
se o secreto é maior ou menor que o palpite. Termine ao acertar ou ao usar as 5 tentativas.

```python
NUMERO_SECRETO = 37
MAX_TENTATIVAS = 5

acertou = 0

for tentativa in range(1, MAX_TENTATIVAS + 1):
    palpite = int(input(f"Tentativa {tentativa} de {MAX_TENTATIVAS}. Digite seu palpite: "))

    if palpite == NUMERO_SECRETO:
        print("Parabéns! Você acertou o número secreto!")
        acertou = 1
        break
    elif palpite < NUMERO_SECRETO:
        print("O número secreto é MAIOR que o seu palpite.")
    else:
        print("O número secreto é MENOR que o seu palpite.")

if acertou == 0:
    print(f"\nSuas tentativas acabaram! O número secreto era {NUMERO_SECRETO}.")
```

##15. Desafio final - caixa de atendimento
Um caixa inicia o dia com R$ 200,00. Leia operações até que o código 0 seja informado. Código 1
= entrada de dinheiro; código 2 = retirada. Para cada operação válida, leia o valor. Uma retirada
só é permitida se houver saldo suficiente. Ao final, informe: saldo final, quantidade de entradas,
quantidade de retiradas realizadas, quantidade de retiradas recusadas, total que entrou e total
que saiu.

```python
saldo = 200.0
entradas_qtd = 0
retiradas_qtd = 0
recusadas_qtd = 0
total_entrou = 0.0
total_saiu = 0.0

codigo = -1

while codigo != 0:
    print(f"\nSaldo atual: R$ {saldo:.2f}")
    codigo = int(input("Informe o código da operação (1 = Entrada, 2 = Retirada, 0 = Sair): "))

    if codigo == 1:
        valor = float(input("Digite o valor da entrada: R$ "))
        if valor > 0:
            saldo += valor
            total_entrou += valor
            entradas_qtd += 1
            print("Entrada realizada com sucesso!")
        else:
            print("Valor inválido!")

    elif codigo == 2:
        valor = float(input("Digite o valor da retirada: R$ "))
        if valor > 0:
            if valor <= saldo:
                saldo -= valor
                total_saiu += valor
                retiradas_qtd += 1
                print("Retirada realizada com sucesso!")
            else:
                recusadas_qtd += 1
                print("Retirada recusada: Saldo insuficiente!")
        else:
            print("Valor inválido!")

    elif codigo != 0:
        print("Código inválido! Tente novamente.")

print("        RESUMO DO DIA - CAIXA       ")

print(f"Saldo final: R$ {saldo:.2f}")
print(f"Quantidade de entradas: {entradas_qtd}")
print(f"Quantidade de retiradas realizadas: {retiradas_qtd}")
print(f"Quantidade de retiradas recusadas: {recusadas_qtd}")
print(f"Total que entrou: R$ {total_entrou:.2f}")
print(f"Total que saiu: R$ {total_saiu:.2f}")
```
